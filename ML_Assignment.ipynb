```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.impute import  SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, classification_report, accuracy_score, precision_score, recall_score, f1_score, mean_squared_error, r2_score
from sklearn.linear_model import LassoCV, RidgeCV


pd.set_option('display.max_column', None)
```

# Assignment 1

House Prine prediction using linaer regression

```python
df = pd.read_csv("/content/drive/MyDrive/AI Engineer Course by Apna College/datasets/HousePricePrediction.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
nums_imputer = SimpleImputer(strategy="mean")
nums_cols = ['BsmtFinSF2', 'TotalBsmtSF', 'SalePrice']
df[nums_cols] = nums_imputer.fit_transform(df[nums_cols])

cat_imputer = SimpleImputer(strategy="most_frequent")
cat_cols = ['MSZoning', 'Exterior1st']
df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])
```

```python
df.isnull().sum()
```

```python
X = df.drop('SalePrice', axis=1)
y = df['SalePrice']

X = pd.get_dummies(X, columns=["MSZoning", "Exterior1st", "LotConfig", "BldgType"], drop_first=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=43)
```

```python
X['BldgType'].unique()
#
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lr_model = LinearRegression()

lr_model.fit(X_train_scaled, y_train)
```

```python
y_pred = lr_model.predict(X_test_scaled)
```

```python
r2 = r2_score(y_test, y_pred)
print("r-squared:", r2)

n = X_test.shape[0]
p = X_test.shape[1]

adjusted_r2 = 1 - ((1-r2) * (n-1) / (n-p-1))
print("adjusted r^2:", adjusted_r2)
```

# Assignment 2



```python
df = pd.read_csv('/content/drive/MyDrive/AI Engineer Course by Apna College/datasets/employee_turnover.csv')
df.head()
```

```python
df.info()
```

```python
X = df.drop('Employee_Turnover', axis=1)
y = df['Employee_Turnover']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=43)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lr_model = LinearRegression()

lr_model.fit(X_train_scaled, y_train)
```

```python
y_pred = lr_model.predict(X_test_scaled)
```

```python
mse = mean_squared_error(y_test, y_pred)
print("mse:", mse)
```

```python
a = [0.001, 0.01, 0.1, 1, 2, 3, 5, 6, 12, 23, 45, 90, 100]

lasso_cv_model = LassoCV(alphas=a, cv=5, max_iter=1000, random_state=34)

lasso_cv_model.fit(X_train_scaled, y_train)

y_pred_lasso_cv = lasso_cv_model.predict(X_test_scaled)
```

```python
print("The best alphe value for lassoCV: ", lasso_cv_model.alpha_)


mse = mean_squared_error(y_test, y_pred_lasso_cv)
print("mse:", mse)

r2 = r2_score(y_test, y_pred_lasso_cv)
print("r2 score:", r2)
```

```python
a = [0.001, 0.01, 0.1, 1, 2, 3, 5, 6, 12, 14, 13, 20, 21, 22, 19, 23, 45, 90, 100]

regit_cv_model = RidgeCV(alphas=a, cv=5)

regit_cv_model.fit(X_train_scaled, y_train)

y_pred_regdi_cv = regit_cv_model.predict(X_test_scaled)
```

```python
print("The best alphe value for RegdiCV: ", regit_cv_model.alpha_)


mse = mean_squared_error(y_test, y_pred_regdi_cv)
print("mse:", mse)

r2 = r2_score(y_test, y_pred_regdi_cv)
print("r2 score:", r2)
```

The best model is linear regression V as the mse is 0.103 which is less then lassoCV 105 and RegdiC 0.104.
