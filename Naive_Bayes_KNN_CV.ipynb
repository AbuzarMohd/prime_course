```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('/content/drive/MyDrive/AI Engineer Course by Apna College/datasets/heart.csv')
df.head()
```

```python
X = df.drop('target', axis=1)
y = df['target']
```

```python
X.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=45)
```

```python
# Naive Bayes
gnb_model = GaussianNB()
gnb_model.fit(X_train, y_train)
```

```python
y_pred = gnb_model.predict(X_test)
```

```python
#Evaluation

print("The recall score: ", recall_score(y_test, y_pred))
print("The accuracy score: ", accuracy_score(y_test, y_pred))
print("The precision score: ", precision_score(y_test, y_pred))
```

# KNN Model

```python
X = df.drop('target', axis=1)
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=45)

scaler = StandardScaler()
X_tarin_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_tarin_scaled
```

```python
knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(X_tarin_scaled, y_train)
```

```python
y_pred_knn = knn_classifier.predict(X_test_scaled)
#Evaluation for k= 3

print("The recall score: ", recall_score(y_test, y_pred_knn))
print("The accuracy score: ", accuracy_score(y_test, y_pred_knn))
print("The precision score: ", precision_score(y_test, y_pred_knn))
```

```python
knn_classifier = KNeighborsClassifier(n_neighbors=5)
knn_classifier.fit(X_tarin_scaled, y_train)
```

```python
y_pred_knn = knn_classifier.predict(X_test_scaled)
#Evaluation for k= 5

print("The recall score: ", recall_score(y_test, y_pred_knn))
print("The accuracy score: ", accuracy_score(y_test, y_pred_knn))
print("The precision score: ", precision_score(y_test, y_pred_knn))
```

```python
knn_classifier = KNeighborsClassifier(n_neighbors=7)
knn_classifier.fit(X_tarin_scaled, y_train)
```

```python
y_pred_knn = knn_classifier.predict(X_test_scaled)
#Evaluation for k= 7

print("The recall score: ", recall_score(y_test, y_pred_knn))
print("The accuracy score: ", accuracy_score(y_test, y_pred_knn))
print("The precision score: ", precision_score(y_test, y_pred_knn))
```

```python
# Cross Validation for hyperparam tuning using GridSearchCV
from sklearn.model_selection import GridSearchCV

classifier = KNeighborsClassifier()
param_grid = {"n_neighbors": [3,5,7,9,11,13,15,17,19]}

classifierCV = GridSearchCV(classifier, param_grid, cv=5, scoring="recall")

classifierCV.fit(X_tarin_scaled, y_train)

y_pred_CV = classifierCV.predict(X_test_scaled)
#Evaluation for classifierCV

print("The recall score: ", recall_score(y_test, y_pred_CV))
print("The accuracy score: ", accuracy_score(y_test, y_pred_CV))
print("The precision score: ", precision_score(y_test, y_pred_CV))

results =pd.DataFrame(classifierCV.cv_results_)
print(results[['param_n_neighbors', 'mean_test_score']])
print(classifierCV.best_params_)
```

```python
# Pipeline
from sklearn.pipeline import Pipeline
df = pd.read_csv('/content/drive/MyDrive/AI Engineer Course by Apna College/datasets/heart.csv')

X = df.drop('target', axis=1)
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=45)

pipeline = Pipeline([('scaler', StandardScaler()), ('knn', KNeighborsClassifier())])

param_grid = {"knn__n_neighbors": [3,5,7,9,11,13,15,17,19]}

classifierCV = GridSearchCV(pipeline, param_grid, cv=5, scoring="recall")

classifierCV.fit(X_train, y_train)

y_pred_CV = classifierCV.predict(X_test)
#Evaluation for classifierCV

print("The recall score: ", recall_score(y_test, y_pred_CV))
print("The accuracy score: ", accuracy_score(y_test, y_pred_CV))
print("The precision score: ", precision_score(y_test, y_pred_CV))

results =pd.DataFrame(classifierCV.cv_results_)
print(classifierCV.best_params_)
```
